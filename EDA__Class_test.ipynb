Bivariate (2 columns)

```python
category_sales = (
    df.groupby("Category")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

category_sales
```

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)
plt.title("Top 10 Categories by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Category")
plt.show()
```

```python
category_sales = (
    df.groupby('Category')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

# Revenue amount bars par show karein
ax.bar_label(
    ax.containers[0],
    fmt='%.2f',
    padding=5
)

plt.title("Top 10 Categories by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

```python
category_sales = (
    df.groupby('Category')['Revenue']
    .sum()
    .sort_values(ascending=False)
    # .head(10)
)

plt.figure(figsize=(11, 6))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

plt.title("Top 10 Categories by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Category")
plt.show()

print(category_sales)
```
